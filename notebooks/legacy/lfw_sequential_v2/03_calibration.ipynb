먼저 EXECUTE=False로 점검하고 실제 계산 시 True로 바꾸세요. [지표·보존 안내](LFW_PAIR_VERIFICATION.md)

```python
CONFIG_PATH = "configs/experiments/lfw_pair_verification.yaml"
EXECUTE = False
KEEP_RAW_RESULTS = False  # 전체 완료·요약 검증 후 이번 campaign 상세 checkpoint만 정리
OUTPUT_ROOT = None  # None: YAML의 output_root. 다른 드라이브의 절대 경로 지정 가능
MAX_NEW_JOBS_PER_RUN = None  # 최초 실행 점검은 1, 전체 미완료 계산은 None
BLAS_THREADS = 2
MIN_AVAILABLE_RAM_GB = 8.0
MAX_PROCESS_RAM_GB = 16.0
```

```python
from pathlib import Path
import sys
import yaml
from IPython.display import display

PROJECT_ROOT = next(p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
                    if (p / "research").is_dir() and (p / "configs").is_dir())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print("Python:", sys.executable)

from research.experiments.lfw_pair_verification import run_verification
pair_result = run_verification(PROJECT_ROOT, config_path=CONFIG_PATH,
    execute=EXECUTE, keep_raw_results=KEEP_RAW_RESULTS, output_root=OUTPUT_ROOT,
    max_new_jobs=MAX_NEW_JOBS_PER_RUN, blas_threads=BLAS_THREADS,
    minimum_available_gb=MIN_AVAILABLE_RAM_GB, maximum_process_gb=MAX_PROCESS_RAM_GB,
    progress=print)
display(pair_result["coverage"])
display({k: v for k, v in pair_result.items() if k not in ("tables", "coverage", "inventory")})
```

```python
REPORT_DIR = pair_result.get("report_dir")
if REPORT_DIR is not None:
    print("04의 REPORT_DIR에 아래 경로를 복사하세요:")
    print(str(REPORT_DIR))
    print("ChatGPT 업로드 안내:", REPORT_DIR / "START_HERE.md")
    print("집계 ZIP:", REPORT_DIR / "analysis.zip")
else:
    print("보고서가 아직 생성되지 않았습니다. 점검 모드 또는 완료 job이 없는 실행입니다.")
```
